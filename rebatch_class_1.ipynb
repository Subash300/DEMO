```
menu = ["Idly", "Dosa", "Vada", "Pongal", "Upma"]
prices = [20, 50, 30, 40, 35]

print("Welcome to the Cafeteria!")

total_bill = 0.0  # Initialize the total bill

while True:
    # Show menu
    print("\nMenu:")
    for i, (item, price) in enumerate(zip(menu, prices), start=1):
        print(f"{i}. {item} - Rs.{price}")

    # Take the user's choice
    try:
        choice = int(input(f"Enter your choice (1-{len(menu)}): "))
        if 1 <= choice <= len(menu):
            print(f"You ordered: {menu[choice - 1]} - Rs.{prices[choice - 1]}")
            total_bill += prices[choice - 1]
        else:
            print("Invalid choice. Please try again.")
            continue
    except ValueError:
        print("Invalid input. Please enter a number.")
        continue

    # Ask if they want to order more
    if input("Would you like to order something else? (yes/no): ").strip().lower() == "no":
        break

# Print the final bill
print("\nYour Total Bill: Rs.", total_bill)
print("Thank you for your order!")
```

```
# TASK : how do i add like to the comments that are present in the post
import datetime

class Post:
    def __init__(self, user, time, content):
        self.user = user  # Username of the post creator
        self.time = time  # Timestamp of the post
        self.content = content  # Post content
        self.like = 0  # Like count for the post
        self.comments = []  # List to store comments (each as a dictionary)

    def like_post(self):
        """Increase the like count for the post."""
        self.like += 1
        return self.like

    def comment_post(self, user, description):
        """Add a comment with username and description. Likes for each comment start at 0."""
        self.comments.append({"user": user, "description": description, "likes": 0})
        return "Comment added"

    def like_comment(self, index):
        """Increase the like count for a specific comment."""
        if 0 <= index < len(self.comments):
            self.comments[index]["likes"] += 1
            return f"Comment by {self.comments[index]['user']} now has {self.comments[index]['likes']} likes."
        else:
            return "Invalid comment index."

    def display_post(self):
        """Display the post details along with comments and likes."""
        print(f"User: {self.user} | Time: {self.time}")
        print(f"Content: {self.content}")
        print(f"Post Likes: {self.like}")
        print("\nComments:")

        if self.comments:
            for idx, comment in enumerate(self.comments, 1):
                print(f"{idx}. {comment['user']}: {comment['description']} (Likes: {comment['likes']})")
        else:
            print("No comments yet.")

# Example usage:
nirmal = Post("Nirmal", datetime.datetime.now(), "Ellam DS padikiranga")

# Liking the post
nirmal.like_post()

# Adding comments
nirmal.comment_post("Subash", "Good post")
nirmal.comment_post("Rahul", "Very informative!")


```

```
class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no
        self.marks = {}  # Dictionary to store subject and marks

    def add_marks(self, subject, mark):
        self.marks[subject] = mark

    def calculate_average(self):
        if len(self.marks) == 0:
            return 0
        return sum(self.marks.values()) / len(self.marks)

    def display_info(self):
        print(f"Student Name: {self.name}")
        print(f"Roll Number: {self.roll_no}")
        print("Marks:")
        for subject, mark in self.marks.items():
            print(f"  {subject}: {mark}")
        print(f"Average Marks: {self.calculate_average():.2f}")

# Creating an object of Student class
student1 = Student("Alice", 101)

# Adding marks
student1.add_marks("Math", 85)
student1.add_marks("Science", 90)
student1.add_marks("English", 78)

# Displaying student information
student1.display_info()
```

```


def binary_search(sorted_list, target):
   low = 0
   high = len(sorted_list) - 1


   while low <= high:
       mid = (low + high) // 2  # Integer division


       if sorted_list[mid] == target:
           return mid  # Target found at index mid
       elif sorted_list[mid] < target:
           low = mid + 1  # Search in the right half
       else:
           high = mid - 1  # Search in the left half


   return -1  # Target not found




# Example usage:
my_list = [2, 5, 7, 8, 11, 12]
target_value = 21


result_index = binary_search(my_list, target_value)


if result_index != -1:
   print(f"Target {target_value} found at index {result_index}")
else:
   print(f"Target {target_value} not found in the list")
```
